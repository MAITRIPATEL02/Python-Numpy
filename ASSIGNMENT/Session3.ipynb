```python
likes = np.array([120, 250, 340, 450, 560, 670, 780])
print("ndim: ",likes.ndim)
print("shape: ",likes.shape)
print("Size: ",likes.size)
print("dtype: ",likes.dtype)
print("itemsize: ",likes.itemsize)
print("nbytes: ",likes.nbytes)
```

TASK 2

```python
steps = np.array([[110,300],[800,300],[500,250],[700,500],[900,400]])
print("Original array:")
print(steps)
print("Original Shape: ",steps.shape)

print("\nConvert 2D into 1D")
one_d = steps.reshape(10)
print(one_d)
print("Shape: ",one_d.shape)

print("\nConvert 1D into 2D")
two_d = one_d.reshape(5,2)
print(two_d)
print("Shape: ",two_d.shape)
```

TASK 3

```python
prices = np.array([
    120, 150, 200, 180,
    250, 300, 350, 400,
    100, 130, 170, 220
])

print("Using ravel()")
r = prices.reshape(3,4).ravel()
print(r)

print("Using flatten()")
f = prices.reshape(3,4).flatten()
print(f)

print("Using resize()")
prices.resize(3,4)
print(prices)

## ravel() : Converts multidimensional array into 1D and returns a view whenever possible
## flatten() : Converts multidimensional array into 1D and returns a copy
## resize() : changes then original array's shape in place
```

```python
playlist = np.array([[10,20,30],[35,25,15],[12,22,32]])
ans1 = playlist.T
print("Using T:")
print(ans1)

ans2 = np.transpose(playlist)
print("Using np.transpose:")
print(ans2)
```

TASK 5

```python
rating = np.array([4,5,2,5,4,1,4,3,5,2,4,5,3,4,5])
print("Original 1D array:")
print(rating)

print("\nAfter convert 1D array to (3,5) array:")
new_rating = rating.reshape(3,5)
print(new_rating)

print("\nAfter convert 2D array to 1D array using flatten():")
f_rating = new_rating.flatten()
print(f_rating)

# flatten() always creates a separate copy.
# ravel() returns a view whenever possible.
# Use flatten() when you need an independent copy.
# Use ravel() when you want to save memory and don't need a separate copy.
```
