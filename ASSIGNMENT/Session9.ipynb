```python
import numpy as np
```

TASK 1

```python
prices = np.array([299, 499, 799, 0, 1599, -1, 899])

print(prices)
print(prices.dtype)
```

TASK 2

```python
prices = np.array([299, 499, 799, 0, 1599, -1, 899])
valid_price = prices[prices > 0]
print(valid_price)

avg_price = np.mean(valid_price)
print("Average price:", avg_price)

prices[ prices <= 0 ] = avg_price
print(prices)
```

TASK 3

```python
quantities = [2, 1, 3, 4, 2, 1, 5]
bills = prices * quantities
print(bills)
```

TASK 4

```python
print("Minimum Price:", np.min(prices))
print("Maximum Price:", np.max(prices))
print("Average Price:", np.mean(prices))
print("Total Price:", np.sum(prices))
print("Total Bill:", np.sum(bills))
```
